# Practical 1B — Fashion MNIST Fully Connected Neural Network

Extracted from `1B.pdf`. Original displayed outputs omitted; executable Python code retained.

**Required local inputs:** `fashion-mnist_test.csv`, `fashion-mnist_train.csv`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
train = pd.read_csv('LP_IV_Updated Datasets/LP_IV_Updated Datasets/FNN and CNN/Fashion MNIST/fashion-mnist_train.csv')
test = pd.read_csv('LP_IV_Updated Datasets/LP_IV_Updated Datasets/FNN and CNN/Fashion MNIST/fashion-mnist_test.csv')
X_train = train.iloc[:, 1:].values
y_train = train.iloc[:, 0].values
X_test = test.iloc[:, 1:].values
y_test = test.iloc[:, 0].values
X_train = X_train / 255.0
X_test = X_test / 255.0
model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
model.compile(
    optimizer='sgd',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(
    X_train, y_train,
    epochs=11, batch_size=32, validation_split=0.1
)
loss, accuracy = model.evaluate(X_test, y_test)
print('Test Loss:', loss)
print('Test Accuracy:', accuracy * 100, '%')
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'])
plt.title('Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'])
plt.title('Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.show()
```
