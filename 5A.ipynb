**Required local inputs:** `vgg16_weights_tf_dim_ordering_tf_kernels_notop.h5`

```python
# a. Set up the environment
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import VGG16
# b. Load and preprocess the data
# Dataset folder path
data_dir = 'LP_IV_Updated Datasets/LP_IV_Updated Datasets/Object Detection/caltech-101'
# Load training data
train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset='training',
    seed=42,
    image_size=(32, 32),
    batch_size=32
)
# Load validation data
val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset='validation',
    seed=42,
    image_size=(32, 32),
    batch_size=32
)
class_names = train_ds.class_names
num_classes = len(class_names)
print("Number of classes:", num_classes)
print("Classes:", class_names)
# Preprocess images for VGG16
preprocess = tf.keras.applications.vgg16.preprocess_input
train_data = train_ds.map(
    lambda x, y: (preprocess(x), y)
)
val_data = val_ds.map(
    lambda x, y: (preprocess(x), y)
)
# c. Choose the pre-trained VGG16 model
base_model = VGG16(
    weights=None,
    include_top=False,
    input_shape=(32, 32, 3)
)
# Load your downloaded VGG16 weights
base_model.load_weights(
    'LP_IV_Updated Datasets/LP_IV_Updated Datasets/Object Detection/vgg16_weights_tf_dim_ordering_tf_kernels_notop.h5'
)
# d. Freeze base layers and add custom layers
base_model.trainable = False
inputs = Input(shape=(32, 32, 3))
x = base_model(inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
x = Dropout(0.3)(x)
outputs = Dense(num_classes, activation='softmax')(x)
model = Model(inputs, outputs)
# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model.summary()
# e. Train the model
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=5
)
# f. Compute loss
loss, accuracy = model.evaluate(val_data)
print("Validation Loss:", loss)
print("Validation Accuracy:", accuracy * 100, "%")
# g. Evaluate the model
print("Model evaluation completed!")
# h. Visualize the prediction
# Get one batch of original images
for images, labels in val_ds.take(1):
    predictions = model.predict(
        preprocess(images), verbose=0
    )
    for i in range(3):
        predicted_class = np.argmax(predictions[i])
        actual_class = labels[i].numpy()
        plt.figure(figsize=(3, 3))
        plt.imshow(images[i].numpy().astype('uint8'))
        plt.title(
            "Predicted: " + class_names[predicted_class] +
            "\nActual: " + class_names[actual_class]
        )
        plt.axis('off')
        plt.show()
# i. Fine-tune the model
# Unfreeze the VGG16 base model
base_model.trainable = True
# Freeze all layers except the last 4
for layer in base_model.layers[:-4]:
    layer.trainable = False
# Compile with a small learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
# Train again for fine-tuning
history_fine = model.fit(
    train_data,
    validation_data=val_data,
    epochs=3
)
# Final evaluation
final_loss, final_accuracy = model.evaluate(val_data)
print("Final Loss:", final_loss)
print("Final Accuracy:", final_accuracy * 100, "%")
# Save the model
model.save('caltech101_vgg16_model.keras')
print("Model saved successfully!")
```
