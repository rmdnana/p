**Required local inputs:** `twitter_dataset.csv`

```python
# 1. Import necessary packages
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
# 2. Load Twitter dataset
data = pd.read_csv('LP_IV_Updated Datasets/LP_IV_Updated Datasets/CBOW/twitter_dataset.csv')
# Select the correct Text column
tweets = data['Text'].dropna().astype(str)
# 3. Clean and tokenize tweets
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()
sentences = [clean_text(tweet) for tweet in tweets]
sentences = [s for s in sentences if len(s) > 0]
# 4. Train CBOW Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=10
)
print("CBOW Model Training Completed!")
# 5. Convert tweets into document vectors
def get_vector(words):
    vectors = [model.wv[w] for w in words if w in model.wv]
    if len(vectors) == 0:
        return None
    return np.mean(vectors, axis=0)
document_vectors = np.array([
    get_vector(s) for s in sentences
])
# 6. Take user input
query = input("Enter a topic or tweet: ")
query_words = clean_text(query)
query_vector = get_vector(query_words)
# 7. Calculate similarity and recommend tweets
if query_vector is None:
    print("No matching words found in vocabulary. Try another topic.")
else:
    similarity = cosine_similarity(
        [query_vector], document_vectors
    )[0]
    top_indices = np.argsort(similarity)[::-1][:5]
    print("\nRecommended Similar Tweets:")
    for i in top_indices:
        print("\nTweet:", tweets.iloc[i])
        print("Similarity:", round(similarity[i], 3))
# 8. Save the model
model.save('recommendation_cbow.model')
print("\nRecommendation Completed!")
```
