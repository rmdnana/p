# Practical 4C — Document CBOW Word2Vec

Extracted from `4C.pdf`. Original displayed outputs omitted; executable Python code retained.

**Required local inputs:** `document3.txt`

```python
# a. Data preparation
# Import necessary packages
import re
from gensim.models import Word2Vec
# Read the text document
file = open('LP_IV_Updated Datasets/LP_IV_Updated Datasets/CBOW/document3.txt', 'r', encoding='utf-8')
text = file.read()
file.close()
print("Original Text:")
print(text)
# Convert text to lowercase
text = text.lower()
# Remove special characters and numbers
text = re.sub(r'[^a-z\s]', '', text)
# Tokenize the text into words
words = text.split()
print("\nTokenized Words:")
print(words)
# Prepare sentences for Word2Vec
sentences = [words[i:i+5] for i in range(len(words)-4)]
# b. Train Word2Vec CBOW Model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=100
)
print("\nCBOW Model Training Completed!")
# c. Evaluate Results
# Display vocabulary
print("\nVocabulary:")
print(model.wv.index_to_key)
# Display word embedding
word = 'influenza'
print("\nWord Embedding of", word)
print(model.wv[word])
# Find similar words
print("\nSimilar Words:")
print(model.wv.most_similar(word, topn=5))
# Check similarity between two words
print("\nSimilarity between COVID and Influenza:")
print(model.wv.similarity('covid', 'influenza'))
# Save the trained model
model.save('document3_cbow.model')
print("\nModel saved successfully!")
```
