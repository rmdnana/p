**Required local inputs:** `twitter_dataset.csv`

```python
# a. Import necessary packages
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
# Load Twitter dataset
data = pd.read_csv('LP_IV_Updated Datasets/LP_IV_Updated Datasets/CBOW/twitter_dataset.csv')
# Display first five rows
print(data.head())
# Display column names
print(data.columns)
# Select tweet column
tweets = data['Text'].dropna().astype(str)
# Clean and tokenize tweets
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    return words
sentences = [clean_text(tweet) for tweet in tweets]
# Remove empty sentences
sentences = [s for s in sentences if len(s) > 0]
print("Total tweets:", len(sentences))
print("Sample tokens:", sentences[0])
# b. Train Word2Vec CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    workers=4,
    epochs=10
)
# Save the trained model
model.save("twitter_cbow.model")
print("CBOW model training completed!")
# c. Evaluate the results
# Display vocabulary
print("Vocabulary size:", len(model.wv))
# Display word embedding
word = list(model.wv.index_to_key)[0]
print("Selected word:", word)
print("Word embedding:", model.wv[word])
# Find similar words
print("Similar words:")
print(model.wv.most_similar(word, topn=5))
```
